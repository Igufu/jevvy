## Model
We'll use Qwen3 0.6B, it has scaled dot-product attention so it's easier to manipulate. In the future it would be good to use a linear attention model.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, Qwen3Model

model_name = "Qwen/Qwen3-0.6B"
```

```python
model = Qwen3Model.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
model
```

```python
prompt = "Junten tokens y aguaio"
input_ids = tokenizer.encode(prompt, return_tensors="pt")

print(input_ids)

print(f'{tokenizer.decode(input_ids[0][0])}{tokenizer.decode(input_ids[0][5])}')
```
